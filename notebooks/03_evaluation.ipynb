# Evaluación del Modelo de Churn

Este notebook carga el modelo entrenado y realiza una evaluación detallada de su rendimiento en el conjunto de prueba.

```python
import pandas as pd
import joblib
import json
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
import sys
sys.path.append('../src')

from data_prep import load_data, split_data
from config import MODEL_PATH, METADATA_PATH
```

## 1. Cargar Modelo, Datos y Metadata

```python
# Cargar el modelo entrenado
model = joblib.load(f'../{MODEL_PATH}')
print(f"Modelo cargado desde {MODEL_PATH}")

# Cargar la metadata
with open(f'../{METADATA_PATH}') as f:
    metadata = json.load(f)
print("Metadata cargada:")
print(json.dumps(metadata, indent=4))

# Cargar y dividir los datos para obtener el conjunto de prueba
df = load_data()
_, X_test, _, y_test = split_data(df)
print(f"\nConjunto de prueba cargado con {len(X_test)} muestras.")
```

## 2. Realizar Predicciones

```python
y_pred = model.predict(X_test)
```

## 3. Matriz de Confusión

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['No Churn', 'Churn'], yticklabels=['No Churn', 'Churn'])
plt.xlabel('Predicción')
plt.ylabel('Verdadero')
plt.title('Matriz de Confusión')
plt.show()
```

## 4. Reporte de Clasificación

```python
print(classification_report(y_test, y_pred, target_names=['No Churn', 'Churn']))
```
