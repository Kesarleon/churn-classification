# Exploratory Data Analysis (EDA) - Churn Classification

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

## 1. Cargar datos

```python
df = pd.read_csv("../data/churn_data.csv")
df.head()
```

## 2. Descripción general

```python
print(df.info())
print("\n")
print(df.describe())
print("\n")
print(df["contract_type"].value_counts())
print("\n")
print(df["internet_service"].value_counts())
print("\n")
print(df["churn"].value_counts(normalize=True))
```

## 3. Distribución de la variable target

```python
sns.countplot(data=df, x="churn", palette="Set2")
plt.title("Distribución de Churn (0 = No, 1 = Sí)")
plt.show()
```

## 4. Variables numéricas vs churn

```python
num_vars = ["age", "tenure", "monthly_charges"]

for var in num_vars:
    sns.boxplot(data=df, x="churn", y=var, palette="Set2")
    plt.title(f"{var} vs churn")
    plt.show()
```

## 5. Variables categóricas vs churn

```python
cat_vars = ["contract_type", "internet_service"]

for var in cat_vars:
    churn_rate = df.groupby(var)["churn"].mean().sort_values(ascending=False)
    churn_rate.plot(kind="bar", color="salmon")
    plt.title(f"Tasa de churn por {var}")
    plt.ylabel("Churn Rate")
    plt.show()
```

## 6. Correlación numérica

```python
corr = df[["age", "tenure", "monthly_charges", "churn"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Matriz de correlación")
plt.show()
```

## 7. Insight final

```python
# Ejemplo de narrativa automática
insights = []
if df.groupby("contract_type")["churn"].mean()["Month-to-month"] > 0.4:
    insights.append("Clientes con contrato 'Month-to-month' presentan alto churn (>40%).")
if df["tenure"].mean() < 20:
    insights.append("Clientes con poca antigüedad tienden a abandonar más rápido.")
if df["monthly_charges"].mean() > 70:
    insights.append("Cargos mensuales altos parecen correlacionar con mayor churn.")

print("🔎 Insights preliminares:")
for i in insights:
    print("-", i)
```
